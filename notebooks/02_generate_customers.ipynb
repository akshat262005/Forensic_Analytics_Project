```python
import pandas as pd
import random
from faker import Faker

fake = Faker("en_IN")

random.seed(42)
```

```python
cities_df = pd.read_csv("cities.csv")

cities_df.head()
```

```python
occupations = {
    "Student": (10000, 30000),
    "Engineer": (50000, 120000),
    "Doctor": (80000, 300000),
    "Teacher": (30000, 70000),
    "Business": (100000, 500000),
    "Government Employee": (40000, 100000),
    "Lawyer": (70000, 250000),
    "Freelancer": (20000, 150000),
    "Retired": (15000, 50000),
    "Shop Owner": (30000, 150000)
}
```

```python
transaction_preferences = {
    "Student": ["UPI", "Shopping", "Food", "Recharge"],
    "Engineer": ["UPI", "Card", "Fuel", "Shopping"],
    "Doctor": ["Transfer", "Card", "Investment"],
    "Teacher": ["UPI", "ATM", "Bills"],
    "Business": ["Transfer", "NEFT", "RTGS", "Salary"],
    "Government Employee": ["UPI", "Bills", "ATM"],
    "Lawyer": ["Transfer", "Card"],
    "Freelancer": ["UPI", "Transfer"],
    "Retired": ["ATM", "Bills"],
    "Shop Owner": ["UPI", "Transfer", "Cash Deposit"]
}
```

```python
num_customers = 5000

customers = []

for i in range(1, num_customers + 1):

    city = cities_df.sample(1).iloc[0]

    occupation = random.choice(list(occupations.keys()))

    income = random.randint(
        occupations[occupation][0],
        occupations[occupation][1]
    )

    age = random.randint(18, 70)

    gender = random.choice(["Male", "Female"])

    # Only 5% suspicious customers
    risk = random.choices(
        ["Normal", "Suspicious"],
        weights=[95,5]
    )[0]

    customers.append({

        "customer_id": f"CUST{i:05d}",

        "name": fake.name(),

        "age": age,

        "gender": gender,

        "occupation": occupation,

        "monthly_income": income,

        "city_id": city["city_id"],

        "city": city["city"],

        "state": city["state"],

        "country": city["country"],

        "home_latitude": city["latitude"],

        "home_longitude": city["longitude"],

        "preferred_transactions":
            ", ".join(transaction_preferences[occupation]),

        "avg_transactions_per_day":
            random.randint(2,15),

        "risk_label": risk

    })
```

```python
customers_df = pd.DataFrame(customers)

customers_df.head()
```
