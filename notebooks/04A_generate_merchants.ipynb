```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import random
import numpy as np

random.seed(42)
np.random.seed(42)

PROJECT_PATH = "/content/drive/MyDrive/AML_Project"
```

```python
cities = pd.read_csv(f"{PROJECT_PATH}/dataset/cities.csv")

cities.head()
```

```python
merchant_categories = {

    "Shopping":[
        "Amazon",
        "Flipkart",
        "Myntra",
        "Reliance Retail",
        "DMart",
        "Croma"
    ],

    "Food":[
        "Swiggy",
        "Zomato",
        "Dominos",
        "Pizza Hut",
        "KFC",
        "McDonalds"
    ],

    "Fuel":[
        "Indian Oil",
        "HP Petrol",
        "Bharat Petroleum",
        "Shell"
    ],

    "Bills":[
        "Electricity Board",
        "Water Department",
        "Gas Agency",
        "Internet Provider",
        "Mobile Recharge"
    ],

    "Healthcare":[
        "Apollo Hospital",
        "Fortis",
        "Max Healthcare",
        "Pharmacy"
    ],

    "Travel":[
        "IRCTC",
        "Uber",
        "Ola",
        "IndiGo",
        "Air India"
    ],

    "Entertainment":[
        "Netflix",
        "Amazon Prime",
        "BookMyShow",
        "PVR"
    ]
}
```

```python
merchants = []

merchant_counter = 1

for category, names in merchant_categories.items():

    for city in cities.itertuples():

        # Every city gets 2–5 merchants from every category
        num = random.randint(2,5)

        for _ in range(num):

            merchant_name = random.choice(names)

            merchants.append({

                "merchant_id":f"MER{merchant_counter:05d}",

                "merchant_name":merchant_name,

                "category":category,

                "city_id":city.city_id,

                "city":city.city,

                "state":city.state,

                "country":city.country,

                "latitude":round(
                    city.latitude +
                    random.uniform(-0.04,0.04),
                    6
                ),

                "longitude":round(
                    city.longitude +
                    random.uniform(-0.04,0.04),
                    6
                )

            })

            merchant_counter += 1
```

```python
merchants = pd.DataFrame(merchants)

print("Total Merchants:",len(merchants))

merchants.head()
```

```python
merchants.to_csv(
    f"{PROJECT_PATH}/dataset/merchants.csv",
    index=False
)

print("Merchants Dataset Saved Successfully")
```
