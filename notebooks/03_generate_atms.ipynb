```python
import pandas as pd
import random

random.seed(42)
```

```python
PROJECT_PATH = "/content/drive/MyDrive/AML_Project"
```

```python
cities_df = pd.read_csv("cities.csv")

cities_df.head()
```

```python
banks = [
    "State Bank of India",
    "HDFC Bank",
    "ICICI Bank",
    "Axis Bank",
    "Punjab National Bank",
    "Bank of Baroda",
    "Canara Bank",
    "Union Bank",
    "Indian Bank",
    "Kotak Mahindra Bank"
]
```

```python
num_atms = 200

atms = []

for i in range(1, num_atms + 1):

    city = cities_df.sample(1).iloc[0]

    # Slight random offset around city center
    latitude = city["latitude"] + random.uniform(-0.05, 0.05)
    longitude = city["longitude"] + random.uniform(-0.05, 0.05)

    atm_type = random.choices(
        ["ATM", "Branch"],
        weights=[80,20]
    )[0]

    bank = random.choice(banks)

    atms.append({

        "atm_id": f"ATM{i:04d}",

        "bank_name": bank,

        "atm_type": atm_type,

        "city_id": city["city_id"],

        "city": city["city"],

        "state": city["state"],

        "country": city["country"],

        "latitude": round(latitude,6),

        "longitude": round(longitude,6)

    })
```

```python
atms_df = pd.DataFrame(atms)

atms_df.head()
```

```python
atms_df.to_csv(
    f"{PROJECT_PATH}/dataset/atms.csv",
    index=False
)

print("ATM Dataset Generated Successfully!")

print()

print(atms_df.head())

print()

print("Total ATMs :", len(atms_df))

print()

print(atms_df["atm_type"].value_counts())
```
